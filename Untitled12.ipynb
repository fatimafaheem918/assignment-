```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv("Downloads/pakistani_ecommerce_dataset.csv")
```

```python
print(df.head())
```

```python
print(df.columns.tolist())
```

```python
CATEGORY = "product_category"
CITY = "city"
PAYMENT = "payment_method"
VALUE = "order_value"
```

```python
print("\nQ1: Most popular product category")
print(df[CATEGORY].value_counts())
print("Answer:", df[CATEGORY].value_counts().idxmax())
```

```python
print("\nQ2: Average order value")
print(round(df[VALUE].mean(), 2))
```

```python
print("\nQ3: Orders per city")
print(df[CITY].value_counts())
print("Answer:", df[CITY].value_counts().idxmax())
```

```python
print("\nQ4: Payment methods")
print(df[PAYMENT].value_counts())
print("Answer.", df[PAYMENT].value_counts().idxmax())
```

```python
avg_by_city = df.groupby(CITY)[VALUE].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=avg_by_city.index, y=avg_by_city.values)
plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value (PKR)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv("Downloads/student_lifestyle_dataset (1).csv")
```

```python
df.columns = [
    "student_id",
    "gender",
    "study_hours",
    "sleep_hours",
    "screen_time",
    "internet_usage",
    "daily_transport_cost",
]
```

```python
print(df.head())
print(df.info())
df.info
```

```python
print("\nQ1: Average study hours")
print(round(df["study_hours"].mean(), 2))
```

```python
print("\nQ2: Average sleep hours by gender")
sleep_by_gender = df.groupby("gender")["sleep_hours"].mean().round(2)
print(sleep_by_gender)
print("Answer:", sleep_by_gender.idxmax(), "sleep more on average")
```

```python
print("\nQ3: Correlation between screen time and sleep hours")
corr = df["screen_time"].corr(df["sleep_hours"])
print(round(corr, 3))
```

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(x="screen_time", y="sleep_hours", data=df)
plt.title("Screen Time vs Sleep Hours")
plt.xlabel("Screen Time (hours)")
plt.ylabel("Sleep Hours")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
plt.hist(df["daily_transport_cost"], bins=10, edgecolor="black")
plt.title("Histogram of Daily Transport Cost")
plt.xlabel("Daily Transport Cost (PKR)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
print("\nQ5: Average internet usage by gender")
internet_by_gender = df.groupby("gender")["internet_usage"].mean().round(2)
print(internet_by_gender)

plt.figure(figsize=(6, 5))
sns.barplot(x=internet_by_gender.index, y=internet_by_gender.values)
plt.title("Average Internet Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Internet Usage (hours)")
plt.tight_layout()
plt.show()
```
